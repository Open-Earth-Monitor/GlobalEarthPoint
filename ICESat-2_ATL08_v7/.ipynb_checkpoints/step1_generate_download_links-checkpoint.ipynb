```python
import earthaccess

# This will prompt for your Earthdata login once and save it to ~/.netrc
earthaccess.login()
```

```python
# Search for the ATL08 v007 dataset. 
# It is highly recommended to add temporal or spatial bounds to match your original order.
results = earthaccess.search_data(
    short_name="ATL08",
    version="007",
    # Add your specific date range to avoid pulling the entire dataset
    # temporal=("2026-07-01", "2026-07-31"), 
    count=500000 # Set higher than your expected total (430,404)
)

# Extract the HTTPS download links
links = []
for granule in results:
    links.extend(granule.data_links())

# Save them to a file for your wget script
output_file = "atl08_links.txt"
with open(output_file, "w") as f:
    for link in links:
        f.write(f"{link}\n")

print(f"Successfully saved {len(links)} links to {output_file}")
```
