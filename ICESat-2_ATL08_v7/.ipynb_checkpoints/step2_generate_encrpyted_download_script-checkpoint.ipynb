```python
import re
import yaml
from datetime import datetime

input_filename = "atl08_links.txt"  # Replace with your actual script filename

# Define the timestamp for the output file
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
output_filename = f"wget_list_{timestamp}.sh"

# Load the token from the YAML configuration file
with open("config.yml", 'r') as config_file:
    config = yaml.safe_load(config_file)
    # Adjust the dictionary keys based on your actual config.yml structure
    my_token = config['earthdata_token']

with open(input_filename, 'r') as infile, open(output_filename, 'w') as outfile:
    for line in infile:
        line = line.strip()
        
        # Process every line that starts with http
        if line.startswith("http"):
            # Extract YYYY/MM/DD from the URL structure
            match = re.search(r'/(\d{4})/(\d{2})/(\d{2})/', line)
            
            if match:
                year, month, day = match.groups()
                date_dir = f"{year}.{month}.{day}"
                
                # Format the new wget command with the dynamically loaded Bearer Token
                wget_command = f'wget --header="Authorization: Bearer {my_token}" -nc -P /mnt/gaia/raw/IceSAT/ATL08.007/{date_dir} {line}\n'
                outfile.write(wget_command)

print(f"Extraction complete. Check {output_filename}")
```
